```python
import gensim.downloader as api

model = api.load("glove-wiki-gigaword-50")
print("Model loaded successfully!")
```

```python
word = "ocean"

print("Word:", word)
print(model[word])
print("Number of dimensions:", len(model[word]))
```

```python
# pairs for the table
required = [("cat", "dog"), ("cat", "computer"), ("king", "queen")]

print("Required pairs:")
for w1, w2 in required:
    print(f"  {w1:10} {w2:10} {model.similarity(w1, w2):.4f}")

# Extra pairs
extra = [("coffee", "tea"), ("doctor", "nurse"), ("river", "banana")]

print("\nExtra pairs:")
for w1, w2 in extra:
    print(f"  {w1:10} {w2:10} {model.similarity(w1, w2):.4f}")
```

```python
for w in ["school", "music", "doctor"]:
    print(f"Most similar to '{w}':")
    for similar, score in model.most_similar(w, topn=5):
        print(f"  {similar:15} {score:.4f}")
    print()
```
